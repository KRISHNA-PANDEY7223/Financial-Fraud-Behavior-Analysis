```python
daily_transaction_more_than_20=df[
    (df["daily_transaction_count"] > 20) &
    (df["fraud_flag"] == True)
].shape[0]
daily_transaction_less_than_20=df[
    (df["daily_transaction_count"] < 20) &
    (df["fraud_flag"] == True)
].shape[0]
print(daily_transaction_more_than_20)
print(daily_transaction_less_than_20)
```

```python
plot_df = pd.DataFrame({
    "Category": ["More than 20", "Less than 20"],
    "Fraud Count": [
        daily_transaction_more_than_20,
        daily_transaction_less_than_20
    ]
})

sns.barplot(data=plot_df, x="Category", y="Fraud Count", palette="magma")
plt.title("Fraud Count by Daily Transaction Count")
plt.show()
```

```python
df["anomaly_score"]=df["anomaly_score"]*100
```

```python
df['anomaly_score_cat']=pd.cut(
    df["anomaly_score"],
    bins=[0,20,40,60,80,100]
)

sns.barplot(data=df,x=df["anomaly_score_cat"],y=df["fraud_flag"],palette="magma")
plt.show()
```

```python
df['transfer_frequency_cat']=pd.cut(
    df["transfer_frequency"],
    bins=[0,15,30,45,60,75,100]
)


pd.crosstab(
    df["anomaly_score_cat"],
    df["transfer_frequency_cat"],
    normalize='index'
)*100
```

```python
pd.crosstab(
    [df["anomaly_score_cat"], df["fraud_flag"]],
    df["transfer_frequency_cat"],
    normalize="index"
) * 100
```
